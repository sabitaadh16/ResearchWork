```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

# Load the clean CSV
csv_path = '/content/drive/MyDrive/DataSet/thyus2path_clean.csv'
df = pd.read_csv(csv_path)

# Convert labels to numbers
# Invalid or empty labels become NaN
df['histo_label'] = pd.to_numeric(
    df['histo_label'],
    errors='coerce'
)

# Keep only valid classes 0 and 1
df = df[df['histo_label'].isin([0, 1])].copy()

# Find the batch from the image path
if 'batch' not in df.columns:
    df['batch'] = df['image_path'].apply(
        lambda path: 'batch1_image'
        if 'batch1_image' in str(path)
        else 'batch2_image'
    )

# Create a patient identifier
# This prevents patient-number overlap between batches
df['patient_key'] = (
    df['batch'].astype(str) + '::' +
    df['patient_name'].astype(str)
)

# Create one row per unique patient
patient_data = df[
    ['patient_key', 'histo_label']
].drop_duplicates()

# Check whether each patient has only one label
label_check = patient_data.groupby('patient_key')['histo_label'].nunique()
print('Patients with more than one label:', (label_check > 1).sum())

# Split patients: 70% train and 30% temporary
train_patients, temporary_patients = train_test_split(
    patient_data,
    test_size=0.30,
    random_state=42,
    stratify=patient_data['histo_label']
)

# Split temporary patients equally: 15% validation and 15% test
validation_patients, test_patients = train_test_split(
    temporary_patients,
    test_size=0.50,
    random_state=42,
    stratify=temporary_patients['histo_label']
)

# Create patient-ID sets
train_ids = set(train_patients['patient_key'])
validation_ids = set(validation_patients['patient_key'])
test_ids = set(test_patients['patient_key'])

# Assign every image to its patient's split
def assign_split(patient_id):
    if patient_id in train_ids:
        return 'train'
    elif patient_id in validation_ids:
        return 'validate'
    elif patient_id in test_ids:
        return 'test'
    else:
        return 'unknown'

df['split'] = df['patient_key'].apply(assign_split)

print('\nUnique patients in each split:')
print(df.groupby('split')['patient_key'].nunique())

# Count unique patients by split and histopathology class
patient_split_data = df[
    ['patient_key', 'split', 'histo_label']
].drop_duplicates()

counts = pd.crosstab(
    patient_split_data['split'],
    patient_split_data['histo_label']
)

# Ensure class 0 and class 1 are shown even if one is absent
counts = counts.reindex(columns=[0, 1], fill_value=0)
counts = counts.reindex(
    ['train', 'validate', 'test'],
    fill_value=0
)

print('\nUnique patient count by split and class:')
print(counts)

# Create the bar graph
ax = counts.plot(
    kind='bar',
    figsize=(9, 6),
    color=['steelblue', 'darkorange'],
    width=0.75
)

# Add count labels above the bars
for container in ax.containers:
    ax.bar_label(container, fmt='%d', padding=3)

plt.title('Patient-Wise Split by Histopathology Label')
plt.xlabel('Dataset Split')
plt.ylabel('Number of Unique Patients')
plt.xticks(rotation=0)
plt.legend(
    title='Histo Label',
    labels=['Class 0', 'Class 1']
)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()

# Save the graph to Google Drive
chart_path = '/content/drive/MyDrive/DataSet/patient_wise_split_class_distribution.png'
plt.savefig(chart_path, dpi=300, bbox_inches='tight')

print('\nGraph saved at:')
print(chart_path)

plt.show()
```

```python
train_df = df[df['split'] == 'train'].copy()

validation_df = df[
    df['split'].isin(['validate', 'validation'])
].copy()

test_df = df[df['split'] == 'test'].copy()

print('Training images:', len(train_df))
print('Validation images:', len(validation_df))
print('Testing images:', len(test_df))
```

```python
train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(10),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

validation_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])
```

```python
class ThyroidDataset(Dataset):

    def __init__(self, dataframe, transform=None):
        self.dataframe = dataframe.reset_index(drop=True)
        self.transform = transform

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, index):
        image_path = self.dataframe.loc[index, 'image_path']
        label = int(self.dataframe.loc[index, 'histo_label'])

        # Open the ultrasound image
        image = Image.open(image_path).convert('RGB')

        # Apply resizing and other transformations
        if self.transform is not None:
            image = self.transform(image)

        return image, label
```

```python
train_dataset = ThyroidDataset(
    train_df,
    transform=train_transform
)

validation_dataset = ThyroidDataset(
    validation_df,
    transform=validation_transform
)

test_dataset = ThyroidDataset(
    test_df,
    transform=validation_transform
)

print('Train Dataset:', len(train_dataset))
print('Validation Dataset:', len(validation_dataset))
print('Test Dataset:', len(test_dataset))
```

```python
import torch
import torch.nn as nn
from torchvision import models
from tqdm import tqdm

# Select GPU if available
device = torch.device(
    'cuda' if torch.cuda.is_available() else 'cpu'
)

print('Using device:', device)

# Load pretrained MobileNetV2
model = models.mobilenet_v2(weights='DEFAULT')

# Replace the final layer for two classes: 0 and 1
number_of_features = model.classifier[1].in_features
model.classifier[1] = nn.Linear(number_of_features, 2)

# Move the model to GPU or CPU
model = model.to(device)

print('MobileNetV2 is ready.')
```

```python
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

# Use training labels only
train_labels = train_df['histo_label'].astype(int).values

class_weight_values = compute_class_weight(
    class_weight='balanced',
    classes=np.array([0, 1]),
    y=train_labels
)

class_weights = torch.tensor(
    class_weight_values,
    dtype=torch.float32
).to(device)

print('Class 0 weight:', class_weights[0].item())
print('Class 1 weight:', class_weights[1].item())
```

```python
loss_function = nn.CrossEntropyLoss(
    weight=class_weights
)

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.0001
)
```

```python
number_of_epochs = 10

train_losses = []
val_losses = []
train_accs = []
val_accs =[]

print("Starting Training:")
for epoch in range(number_of_epochs):

    # Training mode
    model.train()

    train_loss = 0.0
    train_correct = 0
    train_total = 0

    for images, labels in tqdm(train_loader):
        images = images.to(device)
        labels = labels.to(device)

        # Remove old corrections
        optimizer.zero_grad()

        # Make predictions
        outputs = model(images)

        # Calculate error
        loss = loss_function(outputs, labels)

        # Learn from the error
        loss.backward()
        optimizer.step()

        # Record results
        train_loss += loss.item()
        predicted_labels = torch.argmax(outputs, dim=1)
        train_correct += (predicted_labels == labels).sum().item()
        train_total += labels.size(0)

    train_accuracy = train_correct / train_total
    average_train_loss = train_loss / len(train_loader)

    # Validation mode
    model.eval()

    validation_loss = 0.0
    validation_correct = 0
    validation_total = 0

    with torch.no_grad():
        for images, labels in tqdm(validation_loader):
            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)
            loss = loss_function(outputs, labels)

            validation_loss += loss.item()
            predicted_labels = torch.argmax(outputs, dim=1)
            validation_correct += (predicted_labels == labels).sum().item()
            validation_total += labels.size(0)

    validation_accuracy = validation_correct / validation_total
    average_validation_loss = validation_loss / len(validation_loader)

    print(
        'Epoch:', epoch + 1,
        '| Train loss:', round(average_train_loss, 4),
        '| Train accuracy:', round(train_accuracy, 4),
        '| Validation loss:', round(average_validation_loss, 4),
        '| Validation accuracy:', round(validation_accuracy, 4)
    )

    train_losses.append(average_train_loss)
    val_losses.append(average_validation_loss)
    train_accs.append(train_accuracy)
    val_accs.append(validation_accuracy)
```

```python
class ThyroidDataset(Dataset):

    def __init__(self, dataframe, transform):
        self.dataframe = dataframe.reset_index(drop=True)
        self.transform = transform

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, index):
        # Get one row from the CSV
        row = self.dataframe.iloc[index]

        # Open the image from Google Drive
        image = Image.open(row['image_path']).convert('RGB')

        # Get the correct answer: 0 or 1
        label = int(row['histo_label'])

        # Resize and convert the image to numbers
        image = self.transform(image)

        return image, label
```

```python
import torch
import pandas as pd
import matplotlib.pyplot as plt

# Number of training rounds
number_of_epochs = 15

# Location for the best model
best_model_path = (
    '/content/drive/MyDrive/DataSet/'
    'best_mobilenetv2_model.pth'
)

# Keep a record of results
history = []

# The lowest validation loss found so far
best_validation_loss = float('inf')

for epoch in range(number_of_epochs):

    # -----------------------------
    # TRAINING
    # -----------------------------
    model.train()

    total_train_loss = 0.0
    correct_train = 0
    total_train = 0

    for images, labels in train_loader:

        # Move data to GPU or CPU
        images = images.to(device)
        labels = labels.to(device)

        # Remove old gradient information
        optimizer.zero_grad()

        # Make predictions
        outputs = model(images)

        # Calculate the error
        loss = loss_function(outputs, labels)

        # Learn from the error
        loss.backward()
        optimizer.step()

        # Store training results
        total_train_loss += loss.item() * labels.size(0)
        predicted_labels = torch.argmax(outputs, dim=1)
        correct_train += (predicted_labels == labels).sum().item()
        total_train += labels.size(0)

    train_loss = total_train_loss / total_train
    train_accuracy = correct_train / total_train

    # -----------------------------
    # VALIDATION
    # -----------------------------
    model.eval()

    total_validate_loss = 0.0
    correct_validate = 0
    total_validate = 0

    with torch.no_grad():
        for images, labels in validate_loader:

            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)
            loss = loss_function(outputs, labels)

            total_validate_loss += loss.item() * labels.size(0)
            predicted_labels = torch.argmax(outputs, dim=1)
            correct_validate += (predicted_labels == labels).sum().item()
            total_validate += labels.size(0)

    validate_loss = total_validate_loss / total_validate
    validate_accuracy = correct_validate / total_validate

    # Store results for this epoch
    history.append({
        'epoch': epoch + 1,
        'train_loss': train_loss,
        'train_accuracy': train_accuracy,
        'validate_loss': validate_loss,
        'validate_accuracy': validate_accuracy
    })

    print(
        'Epoch:', epoch + 1,
        '| Train loss:', round(train_loss, 4),
        '| Train accuracy:', round(train_accuracy, 4),
        '| Validate loss:', round(validate_loss, 4),
        '| Validate accuracy:', round(validate_accuracy, 4)
    )

    # Save the model if validation loss improved
    if validate_loss < best_validation_loss:
        best_validation_loss = validate_loss

        torch.save({
            'epoch': epoch + 1,
            'model_state_dict': model.state_dict(),
            'optimizer_state_dict': optimizer.state_dict(),
            'validation_loss': validate_loss,
            'validation_accuracy': validate_accuracy
        }, best_model_path)

        print('Best model saved.')
```

```python
all_true_labels = []
all_predicted_labels = []
all_probabilities = []

with torch.no_grad():
    for images, labels in test_loader:

        images = images.to(device)
        labels = labels.to(device)

        outputs = model(images)
        probabilities = torch.softmax(outputs, dim=1)
        predicted_labels = torch.argmax(outputs, dim=1)

        all_true_labels.extend(labels.cpu().numpy())
        all_predicted_labels.extend(predicted_labels.cpu().numpy())
        all_probabilities.extend(probabilities[:, 1].cpu().numpy())

print('Number of test predictions:', len(all_true_labels))
```

```python
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    classification_report
)

# Convert predictions to arrays
y_true = all_true_labels
y_pred = all_predicted_labels
y_probability = all_probabilities

# Confusion matrix order is:
# [[true negative, false positive],
#  [false negative, true positive]]
confusion = confusion_matrix(
    y_true,
    y_pred,
    labels=[0, 1]
)

tn, fp, fn, tp = confusion.ravel()

accuracy = accuracy_score(y_true, y_pred)

# Sensitivity means recall for class 1
sensitivity = recall_score(
    y_true,
    y_pred,
    pos_label=1,
    zero_division=0
)

# Specificity means recall for class 0
specificity = recall_score(
    y_true,
    y_pred,
    pos_label=0,
    zero_division=0
)

precision = precision_score(
    y_true,
    y_pred,
    pos_label=1,
    zero_division=0
)

f1 = f1_score(
    y_true,
    y_pred,
    pos_label=1,
    zero_division=0
)

roc_auc = roc_auc_score(
    y_true,
    y_probability
)

pr_auc = average_precision_score(
    y_true,
    y_probability
)

print('Confusion matrix:')
print(confusion)

print('\nAccuracy:', round(accuracy, 4))
print('Sensitivity:', round(sensitivity, 4))
print('Specificity:', round(specificity, 4))
print('Precision:', round(precision, 4))
print('F1-score:', round(f1, 4))
print('ROC-AUC:', round(roc_auc, 4))
print('PR-AUC:', round(pr_auc, 4))

print('\nClassification report:')
print(
    classification_report(
        y_true,
        y_pred,
        labels=[0, 1],
        target_names=['Class 0', 'Class 1'],
        zero_division=0
    )
)
```
